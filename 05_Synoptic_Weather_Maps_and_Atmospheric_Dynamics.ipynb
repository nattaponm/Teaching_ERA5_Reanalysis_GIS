# 1. Synoptic Meteorology คืออะไร?

Synoptic meteorology ศึกษาโครงสร้างบรรยากาศขนาดใหญ่ในช่วงเวลาหนึ่งหรือหลายช่วงเวลา


```text

```

# 2. Pressure Gradient Force

แรงพื้นฐานที่ทำให้อากาศเร่งจากบริเวณ pressure สูงไปต่ำคือ Pressure Gradient Force

$$
\vec{F}_{PGF}
=
-\frac{1}{\rho}
\nabla p
$$

เครื่องหมายลบหมายถึงแรงชี้ไปทาง pressure ที่ต่ำกว่า

บน pressure surface เรามักอ่าน circulation ผ่าน geopotential-height gradient

# 3. Coriolis Force

Coriolis parameter:

$$
f
=
2\Omega\sin\phi
$$

โดย:

- $\Omega$ = angular rotation rate ของโลก
- $\phi$ = latitude

ใน Northern Hemisphere การเคลื่อนที่ของอากาศถูกเบี่ยงไปทางขวาของทิศการเคลื่อนที่

# 4. Geostrophic Wind

ใน large-scale flow ที่ friction มีบทบาทน้อย ลมสามารถเข้าใกล้ geostrophic balance

$$
f\hat{k}\times\vec{V}_g
=
-\frac{1}{\rho}
\nabla p
$$

บน pressure surface:

$$
\vec{V}_g
=
\frac{g}{f}
\hat{k}
\times
\nabla_p Z
$$

ใน Northern Hemisphere geostrophic wind มีแนวโน้มพัดเกือบขนานกับ geopotential-height contours

แต่ ERA5 `u` และ `v` คือ atmospheric wind fields จาก reanalysis ไม่ใช่ geostrophic wind ที่สมบูรณ์ทุกจุด

# 5. Geopotential Height และ Synoptic Pattern

ERA5 `z` เป็น geopotential

แปลงเป็น geopotential height:

$$
Z
=
\frac{\Phi}{g_0}
$$

Geopotential-height contours ใช้ดู large-scale structures เช่น ridge, trough, closed low, closed high และ height gradient

# 6. Relative Vorticity

Relative vorticity ในแนวดิ่งเขียนได้เป็น:

$$
\zeta
=
\frac{\partial v}{\partial x}
-
\frac{\partial u}{\partial y}
$$

ใน Northern Hemisphere:

```text
ζ > 0
→ cyclonic relative rotation

ζ < 0
→ anticyclonic relative rotation
```

แต่:

> **positive vorticity ≠ rising motion โดยตรง**

# 7. Divergence

Horizontal divergence:

$$
\nabla_h\cdot\vec{V}
=
\frac{\partial u}{\partial x}
+
\frac{\partial v}{\partial y}
$$

การตีความ sign:

```text
divergence > 0
→ horizontal outflow tendency

divergence < 0
→ horizontal convergence tendency
```

โดยทั่วไปมีหน่วย $\mathrm{s^{-1}}$

# 8. Divergence กับ Vertical Motion

จาก mass continuity เราสามารถเชื่อม horizontal divergence กับ vertical motion ได้ในเชิงกายภาพ

```text
low-level convergence
        +
upper-level divergence
        ↓
environment favorable for upward mass transport
```

แต่:

> **divergence ที่ pressure level เดียว ≠ direct measurement of vertical velocity**

และ:

> **upper-level divergence เพียงอย่างเดียว ≠ proof of deep convection**

# 9. ทำไม Vorticity และ Divergence มักคูณด้วย $10^5$?

ค่าของ vorticity/divergence ในบรรยากาศมักมีขนาดประมาณ:

$$
10^{-5}\ \mathrm{s^{-1}}
$$

เพื่อให้อ่าน colorbar ง่าย เราจะแสดง:

$$
\zeta_{plot}
=
10^5\zeta
$$

และ:

$$
D_{plot}
=
10^5D
$$

ดังนั้น label บน figure จะใช้หน่วย:

```text
10⁻⁵ s⁻¹
```

# 10. Synoptic Levels ที่ใช้ในบทนี้

### 850 hPa
ใช้ดู lower-tropospheric thermal field, moisture และ low-level circulation

### 500 hPa
ใช้ดู mid-level geopotential structure, trough/ridge และ relative vorticity

### 200 hPa
ใช้ดู upper-level wind, divergence และ large-scale outflow

นี่เป็น framework สำหรับการเรียน ไม่ใช่กฎตายตัว

# 11. Scientific Interpretation Framework

ทุก figure ในบทนี้ควรอ่านด้วย 3 ขั้น:

```text
1. Observation
   สิ่งที่เห็นจริงจาก field

2. Physical interpretation
   กระบวนการที่สอดคล้องกับ pattern

3. Limitation
   สิ่งที่ field นี้ยังพิสูจน์ไม่ได้
```

ตัวอย่าง:

```text
Observation:
Positive relative vorticity is present over a region.

Interpretation:
The flow exhibits cyclonic relative rotation.

Limitation:
This alone does not prove ascent or severe weather.
```

# 18. กำหนดเวลา, Domain และ Vector Density

````python
# CELL 7 — Analysis settings

REFERENCE_TIME = pd.Timestamp(
    "2011-10-02 00:00:00"
)

ANALYSIS_DOMAIN = {
    "lon_min": 80.0,
    "lon_max": 130.0,
    "lat_min": 0.0,
    "lat_max": 35.0,
}

VECTOR_SKIP = 4

available_times = pd.DatetimeIndex(
    pd.to_datetime(
        ds["time"].values
    )
)

if REFERENCE_TIME not in available_times:
    raise ValueError(
        "Reference time is not available."
    )

print(
    "Reference time:",
    REFERENCE_TIME,
)

print(
    "Analysis domain:",
    ANALYSIS_DOMAIN,
)
````

# 19. Spatial Helper Functions

````python
# CELL 8 — Coordinate-aware spatial subset

def coordinate_slice(
    coordinate,
    lower,
    upper,
):
    values = np.asarray(
        coordinate.values
    )

    if values[-1] > values[0]:
        return slice(
            lower,
            upper,
        )

    return slice(
        upper,
        lower,
    )


def subset_domain(
    data_array,
    domain,
):
    subset = data_array.sel(
        latitude=coordinate_slice(
            data_array["latitude"],
            domain["lat_min"],
            domain["lat_max"],
        ),
        longitude=coordinate_slice(
            data_array["longitude"],
            domain["lon_min"],
            domain["lon_max"],
        ),
    )

    if (
        subset.sizes["latitude"] == 0
        or subset.sizes["longitude"] == 0
    ):
        raise ValueError(
            "Spatial subset is empty."
        )

    return subset
````

````python
# CELL 9 — Map decoration helper

def decorate_map(
    ax,
    *,
    domain,
):
    ax.coastlines(
        resolution="110m",
        linewidth=0.7,
    )

    ax.add_feature(
        cfeature.BORDERS,
        linewidth=0.5,
    )

    ax.set_extent(
        [
            domain["lon_min"],
            domain["lon_max"],
            domain["lat_min"],
            domain["lat_max"],
        ],
        crs=ccrs.PlateCarree(),
    )

    gl = ax.gridlines(
        draw_labels=True,
        linewidth=0.35,
        alpha=0.45,
        linestyle="--",
    )

    gl.top_labels = False
    gl.right_labels = False

    return ax
````

# 20. Derived-Variable Helper Functions

````python
# CELL 10 — Derived atmospheric variables

def geopotential_height(
    geopotential,
):
    height = (
        geopotential
        / G0
    )

    height.name = (
        "geopotential_height"
    )

    height.attrs.update(
        {
            "long_name": "Geopotential height",
            "units": "m",
        }
    )

    return height


def temperature_celsius(
    temperature_K,
):
    temperature_C = (
        temperature_K
        - 273.15
    )

    temperature_C.name = (
        "temperature_C"
    )

    temperature_C.attrs.update(
        {
            "long_name": "Air temperature",
            "units": "degC",
        }
    )

    return temperature_C
````

# 21. เตรียม Synoptic Fields สำหรับเวลาอ้างอิง

````python
# CELL 11 — Prepare reference-time fields

event = ds.sel(
    time=REFERENCE_TIME
)

HEIGHT = geopotential_height(
    event["z"]
)

TEMP_C = temperature_celsius(
    event["t"]
)

RH = event["r"]
U = event["u"]
V = event["v"]
VORTICITY = event["vo"]
DIVERGENCE = event["d"]

print(
    "Reference-time synoptic fields ready."
)
````

# 22. Figure 05-1 — 850-hPa Thermal and Circulation Map

````python
# CELL 12 — 850-hPa temperature + height + wind

level = 850

temp_850 = subset_domain(
    TEMP_C.sel(
        level=level
    ),
    ANALYSIS_DOMAIN,
)

height_850 = subset_domain(
    HEIGHT.sel(
        level=level
    ),
    ANALYSIS_DOMAIN,
)

u_850 = subset_domain(
    U.sel(
        level=level
    ),
    ANALYSIS_DOMAIN,
)

v_850 = subset_domain(
    V.sel(
        level=level
    ),
    ANALYSIS_DOMAIN,
)

fig = plt.figure(
    figsize=(10.5, 6.8)
)

ax = plt.axes(
    projection=ccrs.PlateCarree()
)

mesh = ax.pcolormesh(
    temp_850["longitude"],
    temp_850["latitude"],
    temp_850,
    transform=ccrs.PlateCarree(),
    shading="auto",
    cmap="coolwarm",
)

contours = ax.contour(
    height_850["longitude"],
    height_850["latitude"],
    height_850,
    levels=10,
    transform=ccrs.PlateCarree(),
    linewidths=0.9,
)

ax.clabel(
    contours,
    inline=True,
    fontsize=7,
    fmt="%.0f",
)

skip = VECTOR_SKIP

ax.quiver(
    u_850["longitude"].values[::skip],
    u_850["latitude"].values[::skip],
    u_850.values[::skip, ::skip],
    v_850.values[::skip, ::skip],
    transform=ccrs.PlateCarree(),
)

decorate_map(
    ax,
    domain=ANALYSIS_DOMAIN,
)

cbar = plt.colorbar(
    mesh,
    ax=ax,
    pad=0.025,
    shrink=0.85,
)

cbar.set_label(
    "Temperature (°C)"
)

ax.set_title(
    "850-hPa Temperature, Geopotential Height and Wind\n"
    "2 October 2011 00 UTC"
)

FIG_05_01 = (
    FIGURE_DIR
    / "05_01_850hPa_temperature_height_wind.png"
)

plt.savefig(
    FIG_05_01,
    dpi=300,
    bbox_inches="tight",
)

plt.show()

print(
    "Saved:",
    FIG_05_01,
)
````

# 26. Figure 05-3 — 500-hPa Relative Vorticity, Height and Wind

````python
# CELL 14 — 500-hPa vorticity + height + wind

level = 500

vort_500 = subset_domain(
    VORTICITY.sel(level=level),
    ANALYSIS_DOMAIN,
)

height_500 = subset_domain(
    HEIGHT.sel(level=level),
    ANALYSIS_DOMAIN,
)

u_500 = subset_domain(
    U.sel(level=level),
    ANALYSIS_DOMAIN,
)

v_500 = subset_domain(
    V.sel(level=level),
    ANALYSIS_DOMAIN,
)

vort_scaled = vort_500 * 1.0e5

vort_abs_max = float(
    np.nanpercentile(
        np.abs(vort_scaled.values),
        98,
    )
)

vort_abs_max = max(
    vort_abs_max,
    1.0,
)

norm_vort = TwoSlopeNorm(
    vmin=-vort_abs_max,
    vcenter=0.0,
    vmax=vort_abs_max,
)

fig = plt.figure(
    figsize=(10.5, 6.8)
)

ax = plt.axes(
    projection=ccrs.PlateCarree()
)

mesh = ax.pcolormesh(
    vort_scaled["longitude"],
    vort_scaled["latitude"],
    vort_scaled,
    transform=ccrs.PlateCarree(),
    shading="auto",
    cmap="RdBu_r",
    norm=norm_vort,
)

contours = ax.contour(
    height_500["longitude"],
    height_500["latitude"],
    height_500,
    levels=10,
    transform=ccrs.PlateCarree(),
    linewidths=0.9,
)

ax.clabel(
    contours,
    inline=True,
    fontsize=7,
    fmt="%.0f",
)

skip = VECTOR_SKIP

ax.quiver(
    u_500["longitude"].values[::skip],
    u_500["latitude"].values[::skip],
    u_500.values[::skip, ::skip],
    v_500.values[::skip, ::skip],
    transform=ccrs.PlateCarree(),
)

decorate_map(
    ax,
    domain=ANALYSIS_DOMAIN,
)

cbar = plt.colorbar(
    mesh,
    ax=ax,
    pad=0.025,
    shrink=0.85,
)

cbar.set_label(
    "Relative Vorticity (10⁻⁵ s⁻¹)"
)

ax.set_title(
    "500-hPa Relative Vorticity, Geopotential Height and Wind\n"
    "2 October 2011 00 UTC"
)

FIG_05_03 = (
    FIGURE_DIR
    / "05_03_500hPa_vorticity_height_wind.png"
)

plt.savefig(
    FIG_05_03,
    dpi=300,
    bbox_inches="tight",
)

plt.show()

print(
    "Saved:",
    FIG_05_03,
)
````

# 31. Synoptic Evolution 2–5 October 2011

EVENT dataset มีเพียง:

```text
00 UTC snapshots
2, 3, 4, 5 October 2011
```

ดังนั้น figure ถัดไปแสดง **daily synoptic snapshots**

ไม่ใช่ daily mean และไม่ใช่ full 24-hour evolution

````python
# CELL 17 — Prepare 850-hPa event evolution

event_times = pd.DatetimeIndex(
    pd.to_datetime(
        ds["time"].values
    )
)

evolution_fields = []

for time_value in event_times:
    temp_field = subset_domain(
        temperature_celsius(
            ds["t"].sel(
                time=time_value,
                level=850,
            )
        ),
        ANALYSIS_DOMAIN,
    )

    height_field = subset_domain(
        geopotential_height(
            ds["z"].sel(
                time=time_value,
                level=850,
            )
        ),
        ANALYSIS_DOMAIN,
    )

    u_field = subset_domain(
        ds["u"].sel(
            time=time_value,
            level=850,
        ),
        ANALYSIS_DOMAIN,
    )

    v_field = subset_domain(
        ds["v"].sel(
            time=time_value,
            level=850,
        ),
        ANALYSIS_DOMAIN,
    )

    evolution_fields.append(
        {
            "time": time_value,
            "temperature": temp_field,
            "height": height_field,
            "u": u_field,
            "v": v_field,
        }
    )

print(
    "Prepared snapshots:",
    len(evolution_fields),
)
````

# 32. Figure 05-6 — 850-hPa Synoptic Evolution

````python
# CELL 18 — 850-hPa synoptic evolution

common_temp_min = float(
    min(
        item["temperature"].min()
        for item in evolution_fields
    )
)

common_temp_max = float(
    max(
        item["temperature"].max()
        for item in evolution_fields
    )
)

fig, axes = plt.subplots(
    nrows=2,
    ncols=2,
    figsize=(14, 9),
    subplot_kw={
        "projection": ccrs.PlateCarree()
    },
    constrained_layout=True,
)

axes = axes.ravel()

for ax, item in zip(
    axes,
    evolution_fields,
):
    mesh = ax.pcolormesh(
        item["temperature"]["longitude"],
        item["temperature"]["latitude"],
        item["temperature"],
        transform=ccrs.PlateCarree(),
        shading="auto",
        cmap="coolwarm",
        vmin=common_temp_min,
        vmax=common_temp_max,
    )

    contours = ax.contour(
        item["height"]["longitude"],
        item["height"]["latitude"],
        item["height"],
        levels=8,
        transform=ccrs.PlateCarree(),
        linewidths=0.7,
    )

    ax.clabel(
        contours,
        inline=True,
        fontsize=6,
        fmt="%.0f",
    )

    ax.quiver(
        item["u"]["longitude"].values[::VECTOR_SKIP],
        item["u"]["latitude"].values[::VECTOR_SKIP],
        item["u"].values[::VECTOR_SKIP, ::VECTOR_SKIP],
        item["v"].values[::VECTOR_SKIP, ::VECTOR_SKIP],
        transform=ccrs.PlateCarree(),
    )

    decorate_map(
        ax,
        domain=ANALYSIS_DOMAIN,
    )

    ax.set_title(
        pd.Timestamp(
            item["time"]
        ).strftime(
            "%d %B %Y %H UTC"
        )
    )

cbar = fig.colorbar(
    mesh,
    ax=axes.tolist(),
    orientation="horizontal",
    pad=0.05,
    shrink=0.82,
)

cbar.set_label(
    "850-hPa Temperature (°C)"
)

fig.suptitle(
    "850-hPa Synoptic Evolution\n"
    "Temperature + Geopotential Height + Wind",
    fontsize=15,
)

FIG_05_06 = (
    FIGURE_DIR
    / "05_06_850hPa_synoptic_evolution_2_5Oct2011.png"
)

plt.savefig(
    FIG_05_06,
    dpi=300,
    bbox_inches="tight",
)

plt.show()

print(
    "Saved:",
    FIG_05_06,
)
````

เห็น positive vorticity แล้วสรุปว่ามี rising motion ทันที

เห็น upper-level divergence แล้วสรุปว่าเกิด severe convection แน่นอน

### Mistake 4
ใช้ RH field แทน precipitation

### Mistake 5
ใช้ divergence/vorticity raw values โดยไม่ดู units

### Mistake 6
คูณด้วย $10^5$ แล้วลืมเปลี่ยน label ของ colorbar

### Mistake 7
มองเพียงระดับเดียวแล้วสรุป whole-atmosphere circulation

### Mistake 8
เรียก 00-UTC snapshots ว่า daily means
